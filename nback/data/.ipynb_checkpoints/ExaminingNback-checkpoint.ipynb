# Descriptive Performance Data
## All studies across phases/sessions
### Measures:
#### TotalAcc - total overall accuracy
#### ResponseAcc - accuracy when making a response
#### Dprime - z(hit rate)-z(false alarm rate)
#### Mean Response Time

```r
nback_ops_agg <- nback_ops %>% arrange(study, subjectid, study_event_id = factor(study_event_id, levels = unique(study_event_id)), SessBlockID, TrialNo) %>% filter(!(TrialNo=="NA")) %>% group_by(study, subjectid, study_event_id = factor(study_event_id, levels = unique(study_event_id)), SessBlockID) %>%
    mutate(hit = ifelse(TrialType == "target" & Response == 1, 1, 0),
           miss = ifelse(TrialType == "target" & Response == 0, 1, 0),
           fa = ifelse(TrialType == "empty" & Response == -1, 1, 0),
           cr = ifelse(TrialType == "empty" & Response == 1, 1, 0),
           foil = ifelse(TrialType != "target",
                         case_when(as.numeric(SessBlockID) > 1 & (Stimulus == lag(Stimulus, n=3) | Stimulus == lag(Stimulus, n=1)) ~ 1,
                                   as.numeric(SessBlockID) <= 1 & Stimulus == lag(Stimulus, n=2) ~ 1,
                                   TRUE ~ 0), 0)) %>% 
    summarise(n = n(),
              mrt = mean(ResponseTime[hit==1 | fa==1]),
              mrt_hit = mean(ResponseTime[hit==1]),
              mrt_fa = mean(ResponseTime[fa==1]),
              totalAcc = sum(hit | cr)/sum(hit | miss | fa | cr),
              accFoil = sum(foil[cr==1])/sum(foil),
              accTarget = sum(hit==1)/sum(hit==1 | miss==1),
              nhit = sum(hit),
              nmiss = sum(miss),
              nfa = sum(fa),
              ncr = sum(cr),
              hr = sum(hit)/sum(hit | miss),
              far = sum(fa)/sum(fa | cr),
              hr_adj = (sum(hit)+0.5)/(sum(hit | miss)+1.0),
              far_adj = (sum(fa)+0.5)/(sum(fa | cr)+1.0)
             ) %>% 
    ungroup() %>%
    #group_by(study,study_event_id,SessBlockID) %>%
    mutate(zhr = replace_na(qnorm(hr_adj),0),
           zfar = replace_na(qnorm(far_adj),0),
           dprime = zhr - zfar)
head(nback_ops_agg)

#sig_det_tot$zHR[is.nan(sig_det_tot$zHR)] <- 0
#sig_det_tot$zFAR[is.nan(sig_det_tot$zFAR)] <- 0
#sig_det_tot$Dprime <- (sig_det_tot$zHR - sig_det_tot$zFAR)
#sig_det_tot$Bias <- (sig_det_tot$zHR + sig_det_tot$zFAR)/2

nback_high_agg <- nback_high %>% arrange(study, subjectid, match(study_event_id, c("B1","A1","A2","A3")),  SessBlockID, TrialNo, .by_group = TRUE) %>% filter(!(TrialNo=="NA")) %>% group_by(study, subjectid, study_event_id = factor(study_event_id, levels = unique(study_event_id)), SessBlockID) %>%
    mutate(hit = ifelse(TrialType == "target" & Response == 1, 1, 0),
           miss = ifelse(TrialType == "target" & Response == 0, 1, 0),
           fa = ifelse(TrialType == "empty" & Response == -1, 1, 0),
           cr = ifelse(TrialType == "empty" & Response == 1, 1, 0),
           foil = ifelse(TrialType != "target",
                         case_when(as.numeric(SessBlockID) > 1 & (Stimulus == lag(Stimulus, n=3) | Stimulus == lag(Stimulus, n=1)) ~ 1,
                                   as.numeric(SessBlockID) <= 1 & Stimulus == lag(Stimulus, n=2) ~ 1,
                                   TRUE ~ 0), 0)) %>% 
    summarise(n = n(),
              mrt = mean(ResponseTime[hit==1 | fa==1]),
              mrt_hit = mean(ResponseTime[hit==1]),
              mrt_fa = mean(ResponseTime[fa==1]),
              totalAcc = sum(hit | cr)/sum(hit | miss | fa | cr),
              accFoil = sum(foil[cr==1])/sum(foil),
              accTarget = sum(hit==1)/sum(hit==1 | miss==1),
              nhit = sum(hit),
              nmiss = sum(miss),
              nfa = sum(fa),
              ncr = sum(cr),
              hr = sum(hit)/sum(hit | miss),
              far = sum(fa)/sum(fa | cr),
              hr_adj = (sum(hit)+0.5)/(sum(hit | miss)+1.0),
              far_adj = (sum(fa)+0.5)/(sum(fa | cr)+1.0)
             )  %>% 
    ungroup() %>%
    #group_by(study,study_event_id,SessBlockID) %>%
    mutate(zhr = replace_na(qnorm(hr_adj),0),
           zfar = replace_na(qnorm(far_adj),0),
           dprime = zhr - zfar)
head(nback_high_agg)

nback_agg <- rbind(nback_ops_agg, nback_high_agg)
head(nback_agg)

summarize(nback_agg)
```

```r
## Stability (grouped 1-back and 2-back) (same colors for each participant across DV's) based on mrt

options(repr.plot.width=5, repr.plot.height=5, repr.plot.res=600)

nback_data <- nback_agg %>% select(study, subjectid, study_event_id, SessBlockID, mrt, dprime) %>% 
    melt(id.vars = c("study", "subjectid", "study_event_id", "SessBlockID")) %>%
    filter(!(study_event_id %in% c("Day 5","Day 9"))) %>% 
    mutate(phase = case_when(study_event_id %in% c("Day 0","Day 2","Day 4","Day 6","Day 8","Day 10","Day 13","2","4","6","B1") ~ "Baseline",
                             study_event_id %in% c("Day 15","Day 20","Day 22","Day 27","Day 29","Day 34","Day 41","12","14","17","22","A1","A2","A3") ~ "Stress",
                             TRUE ~ "Recovery"))

nback_data$phase <- factor(nback_data$phase, levels = c("Baseline","Stress","Recovery"))

nback_data <- na.omit(nback_data)

#head(nback_data)

nback_data <- nback_data %>% group_by(study, subjectid, phase, variable) %>% filter(subjectid != "6") %>%
    summarize(avg = mean(value, na.rm=TRUE),
              .groups = 'drop') %>%
    group_by(study) %>% mutate(nphases = n_distinct(phase)) %>%
    group_by(study, subjectid) %>% 
    filter(n_distinct(phase) == nphases) %>%
    complete(phase, variable) %>% 
    group_by(study,subjectid,variable) %>%
    mutate(avg,
           Robustness = if_else(variable=="mrt",avg[phase=="Baseline"]-avg[phase=="Stress"],avg[phase=="Stress"]-avg[phase=="Baseline"]),
           Resilience = if_else(variable=="mrt",avg[phase=="Stress"]-avg[phase=="Recovery"],avg[phase=="Recovery"]-avg[phase=="Stress"])) %>%
    group_by(study,subjectid) %>%
    mutate(Stability = ifelse(study=="HIGH",
                              abs(replace_na(avg[variable=="mrt" & phase=="Baseline"]-avg[variable=="mrt" & phase=="Stress"],0)),
                              abs(replace_na(avg[variable=="mrt" & phase=="Baseline"]-avg[variable=="mrt" & phase=="Stress"],0))+abs(replace_na(avg[variable=="mrt" & phase=="Recovery"]-avg[variable=="mrt" & phase=="Stress"],0)))) %>%
    group_by(study,variable) %>%
    mutate(pStability = 1-percent_rank(Stability),
           pRobustness = percent_rank(Robustness),
           pResilience = percent_rank(Resilience))

#subset(nback_data,study=="OPSII")
head(nback_data)

na.omit(dcast(nback_data, study+subjectid+phase~variable, value.var="avg")) %>% group_by(study) %>% summarize(corr = cor(mrt, dprime))

nback_avg_data <- nback_data %>% group_by(study,phase,variable) %>% summarize(mean=mean(avg, na.rm=TRUE),
                                                   .groups = 'drop')
#head(avg_rt_data)
#comined 1-back and 2-back
rr_plot <- ggplot(data = nback_data, aes(x=phase, y=avg, group=subjectid,color=pStability)) + labs(x="Phase",y="Mean Reaction Time (ms)") + labs(color="Stability\nRank\nPercentile") +
        geom_line(linewidth=0.65, alpha=0.7) + geom_point(size=1.0, alpha=0.8)+
        geom_line(data = nback_avg_data, linewidth=1.0, aes(x=phase, y=mean, group=1), colour="#F100F1FF") +
        theme_bw() + ggtitle("Heterogeneity in Nback") +
        scale_color_viridis(labels = scales::percent) +
        facet_grid2(variable ~ study, scales = "free", switch = "y", labeller = labeller(variable = c(`mrt` = "Mean RT", `dprime` = "d prime"))) + 
        #scale_x_discrete(guide = guide_axis(angle = 60)) +
        #scale_color_manual(values = c("#D55E00", "#0072B2", "#B4AB32")) + 
        theme(text = element_text(size=10, face="bold"),
              plot.title = element_text(size=11, hjust = 0.5),
              axis.text.x = element_text(size=7,angle = 55, hjust = 1),
              axis.title.x = element_text(size=11, margin=margin(5,0,0,0)),
              axis.text.y = element_text(size=7, margin=margin(0,2,0,0)),
              axis.title.y = element_blank(), #element_text(size=11, margin=margin(0,0,0,0),vjust = -16),
              legend.text = element_text(size=9, face="plain"),
              legend.margin = margin(0.25,0,0,0, unit="cm"),
              legend.spacing.y = unit(0.1, 'cm'),
              legend.key.height = unit(1,"line"),
              strip.text.x = element_text(margin = margin(.1, 0, .1, 0, "cm")),
              strip.text.y = element_text(margin = margin(0, .1, 0, .1, "cm")),
              strip.placement.y = "outside"
             )
rr_plot
```

```r
human_ops2_baseline <- subset(nback_ops, study == "OPSII" & SessBlockID == 1 & SessionID <= 6)

human_data <- human_ops2_baseline %>% group_by(subjectid) %>% 
    mutate(hit = ifelse(TrialType == "target" & Response == 1, 1, 0),
           miss = ifelse(TrialType == "target" & Response == 0, 1, 0),
           fa = ifelse(TrialType == "empty" & Response == -1, 1, 0),
           cr = ifelse(TrialType == "empty" & Response == 1, 1, 0)) %>%    
    summarize(mrt = mean(ResponseTime[TrialType=="target"&Response==1]),
              n_hit = sum(hit),
              n_miss = sum(miss),
              n_fa = sum(fa),
              n_cr = sum(cr)) %>%
    mutate(hr = n_hit/(n_hit + n_miss),
           far = n_fa/(n_fa + n_cr),
           acc = (n_hit+n_cr)/(n_hit+n_cr+n_fa+n_miss))

human_data$agent <- "Human"
colnames(human_data)[colnames(human_data) == 'subjectid'] <- 'Subject'
head(human_data)
```
