```r
unique(nback_ops2_data_cleaned$subjectid)
length(unique(nback_ops2_data_cleaned$subjectid))


    
nback_ops2_avg_data <- nback_ops2_data_cleaned %>% arrange(subjectid, study_event_id, SessBlockID, TrialNo) %>% 
    filter(!(TrialNo=="NA")) %>% group_by(subjectid, study_event_id, SessBlockID) %>%
    mutate(hit = ifelse(TrialType == "target" & Response == 1, 1, 0),
           miss = ifelse(TrialType == "target" & Response == 0, 1, 0),
           fa = ifelse(TrialType == "empty" & Response == -1, 1, 0),
           cr = ifelse(TrialType == "empty" & Response == 1, 1, 0),
           foil = ifelse(TrialType != "target",
                         case_when(as.numeric(SessBlockID) > 1 & (Stimulus == lag(Stimulus, n=3) | Stimulus == lag(Stimulus, n=1)) ~ 1,
                                   as.numeric(SessBlockID) <= 1 & Stimulus == lag(Stimulus, n=2) ~ 1,
                                   TRUE ~ 0), 0)) %>%
    group_by(subjectid) %>% filter(n_distinct(phase) == 3) %>%
    group_by(subjectid,phase,SessBlockID) %>%
    summarise(n = n(),
                  mrt = mean(ResponseTime[hit==1 | fa==1]),
                  mrt_hit = mean(ResponseTime[hit==1]),
                  mrt_fa = mean(ResponseTime[fa==1]),
                  totalAcc = sum(hit | cr)/sum(hit | miss | fa | cr),
                  accFoil = sum(foil[cr==1])/sum(foil),
                  accTarget = sum(hit==1)/sum(hit==1 | miss==1),
                  nhit = sum(hit),
                  nmiss = sum(miss),
                  nfa = sum(fa),
                  ncr = sum(cr),
                  hr = sum(hit)/sum(hit | miss),
                  far = sum(fa)/sum(fa | cr),
                  hr_adj = (sum(hit)+0.5)/(sum(hit | miss)+1.0),
                  far_adj = (sum(fa)+0.5)/(sum(fa | cr)+1.0)
                 ) %>% 
        ungroup() %>%
    mutate(zhr = replace_na(qnorm(hr_adj),0),
           zfar = replace_na(qnorm(far_adj),0),
           dprime = zhr - zfar) %>%
    group_by(subjectid,SessBlockID) %>%
    mutate(mrt, dprime,
           delta_stress = (dprime[phase=="Stress"]-dprime[phase=="Baseline"]),
           delta_recovery = (dprime[phase=="Recovery"]-dprime[phase=="Stress"]),
           Robustness = (dprime[phase=="Stress"]-dprime[phase=="Baseline"])/dprime[phase=="Baseline"], 
           Resilience = (dprime[phase=="Recovery"]-dprime[phase=="Stress"])/abs(dprime[phase=="Baseline"]-dprime[phase=="Stress"]),
           logRobustness = log(abs((dprime[phase=="Stress"]-dprime[phase=="Baseline"])/dprime[phase=="Baseline"]))*((dprime[phase=="Stress"]-dprime[phase=="Baseline"])/abs((dprime[phase=="Stress"]-dprime[phase=="Baseline"]))), 
           logResilience = log(abs((dprime[phase=="Recovery"]-dprime[phase=="Stress"])/abs(dprime[phase=="Baseline"]-dprime[phase=="Stress"])))*((dprime[phase=="Recovery"]-dprime[phase=="Stress"])/abs((dprime[phase=="Recovery"]-dprime[phase=="Stress"]))),
           devRobustness = (dprime[phase=="Stress"]-dprime[phase=="Baseline"])/(abs(dprime[phase=="Stress"]-dprime[phase=="Baseline"])+abs(dprime[phase=="Recovery"]-dprime[phase=="Stress"])), 
           devResilience = (dprime[phase=="Recovery"]-dprime[phase=="Stress"])/(abs(dprime[phase=="Stress"]-dprime[phase=="Baseline"])+abs(dprime[phase=="Recovery"]-dprime[phase=="Stress"])))
head(nback_ops2_avg_data)
```

```r
########################   START CREATE MODEL FUNCTION   ###########################
## Create our model function
## Send message to model as json of key:value pairs
## where key = parameter name and value = parameter value
run_model <- function(params,iterations,fix_procedural=FALSE) {
    #startTime <- Sys.time()
    #cat(sprintf("# Start Time: %s\n",startTime))
    if (fix_procedural) {
        params <- list(dat = pvt_params[1], egs = pvt_params[2], ut = pvt_params[3], iu = pvt_params[4], 
                       utmc = pvt_params[5], utbmc = pvt_params[6], fpdec = pvt_params[7], fpmc = pvt_params[8], fpbmc = pvt_params[9], 
                       fdbmc = params[1], fdc = params[2], fddec = params[3], 
                       ans = params[4], bll = params[5], mp = params[6], lf = params[7], 
                       bouts = bouts, numiterations = iterations)
        } else {
        params <- list(dat = params[1], egs = params[2], ut = params[3], iu = params[4], 
                       utmc = params[5], utbmc = params[6], fpdec = params[7], fpmc = params[8], fpbmc = params[9], 
                       fdbmc = params[10], fdc = params[11], fddec = params[12], 
                       ans = params[13], bll = params[14], mp = params[15], lf = params[16], 
                       bouts = bouts, numiterations = iterations)
        }
    cat(sprintf("# request '%s'\n", toJSON(params)))
    params <- toJSON(params, auto_unbox = TRUE)
    cat(sprintf("# sending '%s'\n", params))

    #cat(sprintf("# Elapsed Time: %s\n",Sys.time() - startTime))
    
    sock <- make.socket(host = HOST, port = PORT, server = FALSE)
    on.exit(close.socket(sock))
    
    # Connect to server and send data
    write.socket(sock, paste(params,"\r\n"))
    cat(sprintf("# connected to port '%s'\n", PORT))

    #cat(sprintf("# Elapsed Time: %s\n",Sys.time() - startTime))
    
    # Receive data from the server and shut down
    #received <- character(0)
    received <- character()
    repeat {
        ss <- read.socket(sock)
        if (ss == "") break
        received <- c(received, ss)
    }
    
    #cat(sprintf("# Elapsed Time: %s\n",Sys.time() - startTime))
    
    received <- paste(received, collapse = "")

    cat(sprintf("# received message of length %s\n", nchar(received)))
    raw_data <<- fromJSON(fromJSON(received))

    #cat(sprintf("# Elapsed Time: %s\n",Sys.time() - startTime))
    
    ############## EXAMPLE received DATA FRAME ###############
    ##########################################################
    ##       iteration	day	    bout	curtime	delay	rt
    ##       <int>	    <int>	<int>	<int>	<int>	<int>
    ## 1     1          2	    3	    4	    5       6
    ## 2     2          3	    4	    5	    6       7
    ##########################################################

    # Specify phases
    raw_data$phase <<- ifelse(as.integer(raw_data$day) <= 8, "Baseline",
                                      ifelse(as.integer(raw_data$day) <= 28, "Stress", "Recovery"))
    raw_data$phase <<- factor(raw_data$phase, c("Baseline","Stress","Recovery"))

    raw_data$day <<- factor(raw_data$day, levels=c("2","4","6","12","14","17","22","27","32","38"))
    #return(processed$freq)
    return(raw_data)
}

#########################   END CREATE MODEL FUNCTION   ############################
```

```r
##############################  BEGIN KDE FUNCTION   ###############################
#get_kde <- function(model_values,human_mean) {
#  kde <- density(model_values, bw="nrd0",from=min(model_values,-5),to=max(model_values,5))
#  approx(kde$x, kde$y, xout = human_mean)$y
#}

get_kde <- function(model_values, human_value, metric = "dprime", eps = 1e-4) {
  if (!is.finite(human_value)) {
    return(NA_real_)
  }
  model_values <- model_values[is.finite(model_values)]
  if (length(model_values) < 2 || sd(model_values) == 0) {
    return(eps)
  }

  if (metric == "dprime") {
    from_val <- min(model_values, -5, na.rm = TRUE)
    to_val   <- max(model_values,  5, na.rm = TRUE)
  } else if (metric == "mrt") {
    from_val <- min(model_values, 0, na.rm = TRUE)
    to_val   <- max(model_values, 3, na.rm = TRUE)
  } else {
    from_val <- min(model_values, na.rm = TRUE)
    to_val   <- max(model_values, na.rm = TRUE)
  }
    
  kde <- density(model_values, bw = "nrd0", from = from_val, to = to_val)
  y <- approx(kde$x, kde$y, xout = human_value, rule = 2)$y
  if (!is.finite(y) || is.na(y)) {
    return(eps)
  }
  max(y, eps)
}

human_data <- nback_ops2_avg_data %>% group_by(phase, SessBlockID) %>%
    summarize(mrt = mean(mrt),
              dprime = mean(dprime))
head(human_data)

ll_fun <- function(params) {
    model_data <- run_model(params,iterations,fix_procedural=fix_procedural)

    model_data <- model_data %>% group_by(iteration, day, bout) %>%
        mutate(hit = ifelse(respType == 1 & resp == 1, 1, 0),
               miss = ifelse(respType == 1 & resp == 0, 1, 0),
               fa = ifelse(respType == 0 & resp == 1, 1, 0),
               cr = ifelse(respType == 0 & resp == 0, 1, 0)) %>%
        group_by(iteration,phase,nback) %>%
        summarise(n = n(),
                  mrt = mean(as.numeric(respTime[respType==1&resp==1])),
                  hr_adj = (sum(hit)+0.5)/(sum(hit | miss)+1.0),
                  far_adj = (sum(fa)+0.5)/(sum(fa | cr)+1.0)) %>% 
            ungroup() %>%
        mutate(zhr = replace_na(qnorm(hr_adj),0),
               zfar = replace_na(qnorm(far_adj),0),
               dprime = zhr - zfar)
    model_data$nback <- factor(model_data$nback, levels = c(1,2))


    
    mle_data <- model_data %>% group_by(iteration,phase,nback) %>%
        rename(SessBlockID = nback) %>%
        left_join(human_data, by = c("phase","SessBlockID")) %>%
        group_by(phase,SessBlockID) %>%
        summarise(ll_mrt = {dens <- get_kde(mrt.x, mrt.y[1], metric = "mrt")
                            if (is.na(dens)) 0 else log(dens)
                                },
                  ll_dprime = {dens <- get_kde(dprime.x, dprime.y[1], metric = "dprime")
                               if (is.na(dens)) 0 else log(dens)
                                   },
                  ll = ll_mrt + ll_dprime,
                  .groups = "drop") %>%
        #summarise(ll_mrt = log(get_kde(mrt.x, mrt.y[1]) + 1e-4),
        #          ll_dprime = log(get_kde(dprime.x, dprime.y[1]) + 1e-4),
        #          ll = ll_mrt + ll_dprime,
        #          .groups = "drop") %>% #kde(meanrrt.x,h=bw.nrd0(meanrrt.x),eval.points=mean(meanrrt.y))$estimate)) %>%
        ungroup()

    sum(mle_data$ll)
}

###############################  END KDE FUNCTION   ################################
```

```r
human_dprime_data <- nback_ops2_data_cleaned %>% arrange(subjectid, study_event_id, SessBlockID, TrialNo) %>% 
    filter(!(TrialNo=="NA")) %>% group_by(subjectid, study_event_id, SessBlockID) %>%
    mutate(hit = ifelse(TrialType == "target" & Response == 1, 1, 0),
           miss = ifelse(TrialType == "target" & Response == 0, 1, 0),
           fa = ifelse(TrialType == "empty" & Response == -1, 1, 0),
           cr = ifelse(TrialType == "empty" & Response == 1, 1, 0),
           foil = ifelse(TrialType != "target",
                         case_when(as.numeric(SessBlockID) > 1 & (Stimulus == lag(Stimulus, n=3) | Stimulus == lag(Stimulus, n=1)) ~ 1,
                                   as.numeric(SessBlockID) <= 1 & Stimulus == lag(Stimulus, n=2) ~ 1,
                                   TRUE ~ 0), 0)) %>%
    summarise(n = n(),
                  mrt = mean(ResponseTime[hit==1 | fa==1]),
                  mrt_hit = mean(ResponseTime[hit==1]),
                  mrt_fa = mean(ResponseTime[fa==1]),
                  totalAcc = sum(hit | cr)/sum(hit | miss | fa | cr),
                  accFoil = sum(foil[cr==1])/sum(foil),
                  accTarget = sum(hit==1)/sum(hit==1 | miss==1),
                  nhit = sum(hit),
                  nmiss = sum(miss),
                  nfa = sum(fa),
                  ncr = sum(cr),
                  hr = sum(hit)/sum(hit | miss),
                  far = sum(fa)/sum(fa | cr),
                  hr_adj = (sum(hit)+0.5)/(sum(hit | miss)+1.0),
                  far_adj = (sum(fa)+0.5)/(sum(fa | cr)+1.0)
                 ) %>% 
        ungroup() %>%
    mutate(zhr = replace_na(qnorm(hr_adj),0),
           zfar = replace_na(qnorm(far_adj),0),
           dprime = zhr - zfar) 
head(human_dprime_data)

human_dprime_data <- human_dprime_data %>%
                rename(day = study_event_id,
                       block = SessBlockID) %>%
                mutate(agent = "human")
head(human_dprime_data)
```

```r
model_data <- nback_model_data %>% group_by(iteration, day, block) %>%
    mutate(hit = ifelse(respType == 1 & resp == 1, 1, 0),
           miss = ifelse(respType == 1 & resp == 0, 1, 0),
           fa = ifelse(respType == 0 & resp == 1, 1, 0),
           cr = ifelse(respType == 0 & resp == 0, 1, 0),
           foil = ifelse(respType != 1,
                         case_when(block > 1 & (stim == lag(stim, n=3) | stim == lag(stim, n=1)) ~ 1,
                                   block <= 1 & stim == lag(stim, n=2) ~ 1,
                                   TRUE ~ 0), 0)) %>% 
    summarise(n = n(),
              mrt = mean(as.numeric(respTime[respType==1&resp==1])),
              mrt_hit = mean(respTime[hit==1]),
              mrt_fa = mean(respTime[fa==1]),
              totalAcc = sum(hit | cr)/sum(hit | miss | fa | cr),
              accFoil = sum(foil[cr==1])/sum(foil),
              accTarget = sum(hit==1)/sum(hit==1 | miss==1),
              nhit = sum(hit),
              nmiss = sum(miss),
              nfa = sum(fa),
              ncr = sum(cr),
              hr = sum(hit)/sum(hit | miss),
              far = sum(fa)/sum(fa | cr),
              hr_adj = (sum(hit)+0.5)/(sum(hit | miss)+1.0),
              far_adj = (sum(fa)+0.5)/(sum(fa | cr)+1.0)
             ) %>% 
    ungroup() %>%
    #group_by(study,study_event_id,SessBlockID) %>%
    mutate(zhr = replace_na(qnorm(hr_adj),0),
           zfar = replace_na(qnorm(far_adj),0),
           dprime = zhr - zfar)
head(model_data)

model_data <- model_data %>%
                rename(subjectid = iteration) %>%
                mutate(agent = "model",
                       subjectid = as.factor(subjectid),
                       block = as.factor(block))
head(model_data)
```

```r
model_data <- nback_model_data %>% group_by(iteration, day, block) %>%
    mutate(hit = ifelse(respType == 1 & resp == 1, 1, 0),
           miss = ifelse(respType == 1 & resp == 0, 1, 0),
           fa = ifelse(respType == 0 & resp == 1, 1, 0),
           cr = ifelse(respType == 0 & resp == 0, 1, 0),
           foil = ifelse(respType != 1,
                         case_when(block > 1 & (stim == lag(stim, n=3) | stim == lag(stim, n=1)) ~ 1,
                                   block <= 1 & stim == lag(stim, n=2) ~ 1,
                                   TRUE ~ 0), 0)) %>% 
    summarise(n = n(),
              mrt = mean(as.numeric(respTime[respType==1&resp==1])),
              mrt_hit = mean(respTime[hit==1]),
              mrt_fa = mean(respTime[fa==1]),
              totalAcc = sum(hit | cr)/sum(hit | miss | fa | cr),
              accFoil = sum(foil[cr==1])/sum(foil),
              accTarget = sum(hit==1)/sum(hit==1 | miss==1),
              nhit = sum(hit),
              nmiss = sum(miss),
              nfa = sum(fa),
              ncr = sum(cr),
              hr = sum(hit)/sum(hit | miss),
              far = sum(fa)/sum(fa | cr),
              hr_adj = (sum(hit)+0.5)/(sum(hit | miss)+1.0),
              far_adj = (sum(fa)+0.5)/(sum(fa | cr)+1.0)
             ) %>% 
    ungroup() %>%
    #group_by(study,study_event_id,SessBlockID) %>%
    mutate(zhr = replace_na(qnorm(hr_adj),0),
           zfar = replace_na(qnorm(far_adj),0),
           dprime = zhr - zfar)
head(model_data)

model_data <- model_data %>%
                rename(subjectid = iteration) %>%
                mutate(agent = "model",
                       subjectid = as.factor(subjectid),
                       block = as.factor(block))
head(model_data)
```

## Model vs Human

```r
human_ops2_baseline <- subset(nback_ops, study == "OPSII" & SessBlockID == 1 & SessionID <= 6)

human_data <- human_ops2_baseline %>% group_by(subjectid) %>% 
    mutate(hit = ifelse(TrialType == "target" & Response == 1, 1, 0),
           miss = ifelse(TrialType == "target" & Response == 0, 1, 0),
           fa = ifelse(TrialType == "empty" & Response == -1, 1, 0),
           cr = ifelse(TrialType == "empty" & Response == 1, 1, 0)) %>%    
    summarize(mrt = mean(ResponseTime[TrialType=="target"&Response==1]),
              n_hit = sum(hit),
              n_miss = sum(miss),
              n_fa = sum(fa),
              n_cr = sum(cr)) %>%
    mutate(hr = n_hit/(n_hit + n_miss),
           far = n_fa/(n_fa + n_cr),
           acc = (n_hit+n_cr)/(n_hit+n_cr+n_fa+n_miss))

human_data$agent <- "Human"
colnames(human_data)[colnames(human_data) == 'subjectid'] <- 'Subject'
head(human_data)
```

```r
model_data_mp1 <- read.csv("nback_model_data_mp1.csv",header=T)
head(model_data_mp1)

model_data_mp1 <- model_data_mp1 %>% group_by(Subject) %>% 
    mutate(hit = ifelse(RespType == 1 & Response == 1, 1, 0),
           miss = ifelse(RespType == 1 & Response == 0, 1, 0),
           fa = ifelse(RespType == 0 & Response == 1, 1, 0),
           cr = ifelse(RespType == 0 & Response == 0, 1, 0)) %>%    
    summarize(mrt = mean(as.numeric(RespTime[RespType==1&Response==1])),
              n_hit = sum(hit),
              n_miss = sum(miss),
              n_fa = sum(fa),
              n_cr = sum(cr)) %>%
    mutate(hr = n_hit/(n_hit + n_miss),
           far = n_fa/(n_fa + n_cr),
           acc = (n_hit+n_cr)/(n_hit+n_cr+n_fa+n_miss))

model_data_mp1$agent <- "Model"
head(model_data_mp1)
```

```r
options(repr.plot.width=6.5, repr.plot.height=3.5, repr.plot.res=600)
data <- rbind(human_data,model_data_mp1)

##Accuracy X Response Time
plot <- ggplot(data = data, aes(x=acc, y=mrt, group=agent,color=agent,shape=agent,alpha=agent)) + labs(x="Overall Accuracy",y="Mean Response Time") + labs(color="Phase") +
        geom_point(size=1.5, stroke=0.9, position=position_jitter()) + 
        scale_alpha_manual(values=c(0.7,0.5),guide="none") +
        scale_shape_manual(values=c(1,1),guide="none") +
        scale_color_manual(values = c("#D55E00", "#0072B2")) +
        guides(color = guide_legend(override.aes = list(shape = c(1,1)))) +
        theme_bw() + ggtitle("Nback OPSII - Human vs Model") +
        theme(text = element_text(size=10, face="bold"),
              plot.title = element_text(size=11, hjust = 0.5),
              axis.text.x = element_text(size=9),
              axis.title.x = element_text(size=11, margin=margin(5,0,0,0)),
              axis.text.y = element_text(size=9, margin=margin(0,2,0,0)),
              axis.title.y = element_text(size=11, margin=margin(0,9,0,0)),
              legend.text = element_text(size=9, face="plain"),
              legend.margin = margin(0.25,0,0,0, unit="cm"),
              legend.spacing.y = unit(0.1, 'cm'),
              legend.key.height = unit(1,"line")
             )

plot

##hr X far
plot <- ggplot(data = data, aes(x=far, y=hr, group=agent,color=agent,shape=agent,alpha=agent)) + labs(x="False Alarm Rate",y="Hit Rate") + labs(color="Phase") +
        geom_point(size=1.5, stroke=0.9, position=position_jitter()) + 
        scale_alpha_manual(values=c(0.7,0.5),guide="none") +
        scale_shape_manual(values=c(1,1),guide="none") +
        scale_color_manual(values = c("#D55E00", "#0072B2")) +
        guides(color = guide_legend(override.aes = list(shape = c(1,1)))) +
        theme_bw() + ggtitle("Nback OPSII - Human vs Model") +
        theme(text = element_text(size=10, face="bold"),
              plot.title = element_text(size=11, hjust = 0.5),
              axis.text.x = element_text(size=9),
              axis.title.x = element_text(size=11, margin=margin(5,0,0,0)),
              axis.text.y = element_text(size=9, margin=margin(0,2,0,0)),
              axis.title.y = element_text(size=11, margin=margin(0,9,0,0)),
              legend.text = element_text(size=9, face="plain"),
              legend.margin = margin(0.25,0,0,0, unit="cm"),
              legend.spacing.y = unit(0.1, 'cm'),
              legend.key.height = unit(1,"line")
             )

plot
```
