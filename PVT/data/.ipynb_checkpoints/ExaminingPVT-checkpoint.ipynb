```r
pvt_data <- pvt_ops_raw
unique(pvt_data$study)

#pid_raw is very messed up
#A00-07-6166 SessionID 33 needs to change to 32
#A00-07-8758 SessionID 33 needs to change to 32
#A00-07-1637 SessionID 37 needs to change to 38
#A00-07-4627 SessionID 39 needs to change to 38
#### or just use 'subjectid' and 'days' which are already corrected
#using 'subjectid', 'days'

#Fix errors with SessionIDs for these two participants: A00-07-2880 & A00-07-4590
pvt_data$SessionID[pvt_data$subjectid=="A00-07-2880" & pvt_data$days=="          29"] <- 29
pvt_data$SessionID[pvt_data$subjectid=="A00-07-4590" & pvt_data$days=="          22"] <- 22

#Fix error with subjectid for this participant: A00-07-288
pvt_data$subjectid[pvt_data$subjectid=="A00-07-2880"] <- "A00-06-2880"

#First need to clean data with duplicate sessions, and then fix messed up days, date1, and date2 fields for ops1
pvt_data <- pvt_data[!duplicated(pvt_data[1:30]), ] ##removes duplicate rows, but leaving those with only incorrectly labeled SessionID's

days <- pvt_data %>%
    group_by(study, subjectid) %>% filter(!is.na(StartDT)) %>% arrange(study, subjectid, SessionID) %>%
    mutate(days = ifelse(study == "OPSI", 
                         as.character(as.numeric(as.Date(StartDT) - as.Date(StartDT[1]))),
                         as.character(as.numeric(as.Date(StartDT) - as.Date(StartDT[1])) + 2))) %>% 
    mutate(date2 = ifelse(study == "OPSI",
                          ifelse(subjectid == "A00-06-0739" & as.numeric(days)>=16 & as.numeric(days)<=42, paste("Day ", as.numeric(days)-1, sep=""),
                                 ifelse(as.numeric(days)==3, "Day 2",
                                        ifelse(as.numeric(days)==7, "Day 6",
                                               ifelse(as.numeric(days)>56, "EOS",
                                                      ifelse(as.numeric(days)>=53, "Day 54", paste("Day ", days, sep="")))))),
                          ifelse(subjectid == "A00-07-2030" | subjectid == "A00-07-4258", paste("Day ", SessionID, sep=""),
                                 ifelse(as.numeric(days)==7, "Day 6",
                                        ifelse(as.numeric(days)==16 | as.numeric(days)==18, "Day 17",
                                               ifelse(as.numeric(days)==21, "Day 22",
                                                      ifelse(as.numeric(days)==26, "Day 27",
                                                             ifelse(as.numeric(days)==31 | as.numeric(days)==33, "Day 32",
                                                                    ifelse(as.numeric(days)>33, "Day 38", paste("Day ", days, sep="")))))))))) %>%
    mutate(date1 = paste(date2, "_", month(as.Date(StartDT)), ".", day(as.Date(StartDT)),".",format(as.Date(StartDT), "%y"), sep="")) %>%
    select(study,subjectid,SessionID,days,date1,date2)
#days

pvt_data <- pvt_data %>% rows_update(days, by = c("study","subjectid","SessionID"))
#pvt_data

#add study_event_id
pvt_data$study_event_id <- ifelse(pvt_data$study=="OPSI", pvt_data$date2, substr(pvt_data$date2,5,nchar(pvt_data$date2)))
#pvt_data$RespTime <- as.double(pvt_data$RespTime)
#pvt_data
#nrow(subset(pvt_data, study_event_id == "ERROR"))
unique(pvt_data$study_event_id)

#Filter agg data from files
pvt_data <- pvt_data %>% filter(RecType=="P" | RecType=="V" | RecType=="T")

#Fix Lapses and Filter invalid Prematures
pvt_data_cleaned <- pvt_data %>% group_by(study, subjectid, study_event_id) %>% 
    mutate(RecType = case_when(RecType=="T" & lead(RecType) == "P" & lead(ElapsedSec)-ElapsedSec<=0.5 ~ "L", TRUE ~ RecType),
           RespTime = ifelse(RecType=="L", 0.5+lead(ElapsedSec)-ElapsedSec, RespTime),
           ElapsedSec = ifelse(RecType=="L", lead(ElapsedSec), ElapsedSec)
          ) %>%
    filter(!(row_number()>1 & RecType == "P" & lag(RecType) == "L" & ElapsedSec >= lag(ElapsedSec) & ElapsedSec <= 0.5 + lag(ElapsedSec)))


pvt_data_cleaned$rt <- ifelse(pvt_data_cleaned$RecType == "T", 1000,
                               ifelse(pvt_data_cleaned$RecType == "P", 0, 
                                      ifelse(pvt_data_cleaned$RespTime < 0.150, 0, pvt_data_cleaned$RespTime * 1000)))

# Specify phases
pvt_data_cleaned$phase <- ifelse(pvt_data_cleaned$study=="OPSI", ifelse(pvt_data_cleaned$study_event_id == "EOS","Recovery",
                                                                        ifelse(as.integer(gsub("[A-Za-z]", "", pvt_data_cleaned$study_event_id)) < 14, "Baseline",
                                                                               ifelse(as.integer(gsub("[A-Za-z]", "", pvt_data_cleaned$study_event_id)) < 42, "Stress", "Recovery"))),
                                 ifelse(as.integer(pvt_data_cleaned$study_event_id) < 8, "Baseline",
                                        ifelse(as.integer(pvt_data_cleaned$study_event_id) < 28, "Stress", "Recovery")))
pvt_data_cleaned$phase <- factor(pvt_data_cleaned$phase, c("Baseline","Stress","Recovery"))

pvt_data_cleaned$study_event_id <- factor(pvt_data_cleaned$study_event_id,c("Day 0","Day 2","Day 4","Day 5","Day 6","Day 8","Day 9","Day 10","Day 13","Day 15","Day 20","Day 22","Day 27","Day 29","Day 34","Day 36","Day 41","Day 54","EOS",
                                                                                   "2","4","6","12","14","17","22","27","32","38"))

pvt_data_cleaned <- pvt_data_cleaned %>% group_by(subjectid, study_event_id) %>%
    filter(sum(!is.na(RespTime)) != 0)


pvt_data_cleaned$study <- as.factor(pvt_data_cleaned$study)
pvt_data_cleaned$subjectid <- as.factor(pvt_data_cleaned$subjectid)
pvt_data_cleaned$SessBlockID <- as.factor(pvt_data_cleaned$SessBlockID)

unique(pvt_data_cleaned$study_event_id)
head(pvt_data_cleaned)
```

# Descriptive Performance Data
## All studies across phases/sessions
### Measures:
#### nvalid = number of valid trials (rt >= 150 ms and < 500 ms)
#### nlapses = number of lapses (rt > 500 ms and less than 1000 ms)
#### nfalsestart = number of false starts (rt < 150 ms)
#### ntimeouts = number of timeouts (no response)
#### nfslapse = number of false starts + lapses
#### ntotalerr = number of false starts + lapses + timeouts
#### mrtime = mean response time of valid and lapse responses(ms)
#### mrtimeslow10 = mean response time of slowest 10% of valid and lapse responses
#### mrtimefast10 = mean response time of fastest 10% of valid and lapse responses
#### meanrrt = mean reciprocal response time [mean(1/rt)]
#### meanrrtslow10 = meanrrt of slowest 10% of valid and lapse responses
#### meanrrtfast10 = meanrrt of fastest 10% of valid and lapse responses

```r
pvt_ops_agg <- pvt_data_cleaned %>% group_by(study,subjectid,phase,study_event_id) %>%
            summarize(npresented = n(),
                      nvalid = sum(RecType=="V" & rt>=150),
                      validp = sum(RecType=="V" & rt>=150)/n(),
                      nfalsestart = sum(RecType=="FS" | (RecType=="V" & rt<150)),
                      nlapses = sum(RecType=="L"),
                      ntimeouts = sum(RecType=="T"),
                      nfslapse = sum(RecType=="FS" | (RecType=="V" & rt<150) | RecType=="L"),
                      nlapseto = sum(RecType=="L" | RecType=="T"),
                      ntotalerr = n() - sum(RecType=="V" & rt>=150),
                      mrtime = mean(rt[(RecType=="V" | RecType=="L") & rt>=150]),
                      variancert = sd(rt[(RecType=="V" | RecType=="L") & rt>=150]),
                      medianrt = quantile(rt[(RecType=="V" | RecType=="L") & rt>=150],0.5,type=1),
                      mrtimeslow10 = mean(rt[(RecType=="V" | RecType=="L") & rt>=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.9)]),
                      mrtimefast10 = mean(rt[(RecType=="V" | RecType=="L") & rt>=150 & rt<=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.1)]),
                      meanrrt = mean(1/(rt[(RecType=="V" | RecType=="L") & rt>=150]/1000)),
                      variancerrt = sd(1/(rt[(RecType=="V" | RecType=="L") & rt>=150]/1000)),
                      medianrrt = quantile(1/(rt[(RecType=="V" | RecType=="L") & rt>=150]/1000),0.5,type=1),
                      meanrrtslow10 = mean(1/(rt[(RecType=="V" | RecType=="L") & rt>=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.9)]/1000)),
                      meanrrtfast10 = mean(1/(rt[(RecType=="V" | RecType=="L") & rt>=150 & rt<=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.1)]/1000)),
                      )
head(pvt_ops_agg)

pvt_sere_agg <- pvt_sere_cleaned %>% group_by(study,subjectid,phase,study_event_id) %>%
            summarize(npresented = n(),
                      nvalid = sum(RecType=="V" & rt>=150),
                      validp = sum(RecType=="V" & rt>=150)/n(),
                      nfalsestart = sum(RecType=="FS" | (RecType=="V" & rt<150)),
                      nlapses = sum(RecType=="L"),
                      ntimeouts = sum(RecType=="T"),
                      nfslapse = sum(RecType=="FS" | (RecType=="V" & rt<150) | RecType=="L"),
                      nlapseto = sum(RecType=="L" | RecType=="T"),
                      ntotalerr = n() - sum(RecType=="V" & rt>=150),
                      mrtime = mean(rt[(RecType=="V" | RecType=="L") & rt>=150]),
                      variancert = sd(rt[(RecType=="V" | RecType=="L") & rt>=150]),
                      medianrt = quantile(rt[(RecType=="V" | RecType=="L") & rt>=150],0.5,type=1),
                      mrtimeslow10 = mean(rt[(RecType=="V" | RecType=="L") & rt>=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.9)]),
                      mrtimefast10 = mean(rt[(RecType=="V" | RecType=="L") & rt>=150 & rt<=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.1)]),
                      meanrrt = mean(1/(rt[(RecType=="V" | RecType=="L") & rt>=150]/1000)),
                      variancerrt = sd(1/(rt[(RecType=="V" | RecType=="L") & rt>=150]/1000)),
                      medianrrt = quantile(1/(rt[(RecType=="V" | RecType=="L") & rt>=150]/1000),0.5,type=1),
                      meanrrtslow10 = mean(1/(rt[(RecType=="V" | RecType=="L") & rt>=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.9)]/1000)),
                      meanrrtfast10 = mean(1/(rt[(RecType=="V" | RecType=="L") & rt>=150 & rt<=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.1)]/1000)),
                      )
head(pvt_sere_agg)

pvt_high_agg <- pvt_high_cleaned %>% group_by(study,subjectid,phase,study_event_id) %>%
            summarize(npresented = n(),
                      nvalid = sum(RecType=="V" & rt>=150),
                      validp = sum(RecType=="V" & rt>=150)/n(),
                      nfalsestart = sum(RecType=="FS" | (RecType=="V" & rt<150)),
                      nlapses = sum(RecType=="L"),
                      ntimeouts = sum(RecType=="T"),
                      nfslapse = sum(RecType=="FS" | (RecType=="V" & rt<150) | RecType=="L"),
                      nlapseto = sum(RecType=="L" | RecType=="T"),
                      ntotalerr = n() - sum(RecType=="V" & rt>=150),
                      mrtime = mean(rt[(RecType=="V" | RecType=="L") & rt>=150]),
                      variancert = sd(rt[(RecType=="V" | RecType=="L") & rt>=150]),
                      medianrt = quantile(rt[(RecType=="V" | RecType=="L") & rt>=150],0.5,type=1),
                      mrtimeslow10 = mean(rt[(RecType=="V" | RecType=="L") & rt>=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.9)]),
                      mrtimefast10 = mean(rt[(RecType=="V" | RecType=="L") & rt>=150 & rt<=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.1)]),
                      meanrrt = mean(1/(rt[(RecType=="V" | RecType=="L") & rt>=150]/1000)),
                      variancerrt = sd(1/(rt[(RecType=="V" | RecType=="L") & rt>=150]/1000)),
                      medianrrt = quantile(1/(rt[(RecType=="V" | RecType=="L") & rt>=150]/1000),0.5,type=1),
                      meanrrtslow10 = mean(1/(rt[(RecType=="V" | RecType=="L") & rt>=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.9)]/1000)),
                      meanrrtfast10 = mean(1/(rt[(RecType=="V" | RecType=="L") & rt>=150 & rt<=quantile(rt[(RecType=="V" | RecType=="L") & rt>=150], 0.1)]/1000)),
                      )
head(pvt_high_agg)
```

## Measures:
### Dprime - z(hit rate)-z(false alarm rate)
### Mean Response Time


```r
## Stability (grouped 1-back and 2-back) (same colors for each participant across DV's) based on mrt

options(repr.plot.width=5, repr.plot.height=5, repr.plot.res=600)

nback_data <- nback_agg %>% select(study, subjectid, study_event_id, SessBlockID, mrt, dprime) %>% 
    melt(id.vars = c("study", "subjectid", "study_event_id", "SessBlockID")) %>%
    filter(!(study_event_id %in% c("Day 5","Day 9"))) %>% 
    mutate(phase = case_when(study_event_id %in% c("Day 0","Day 2","Day 4","Day 6","Day 8","Day 10","Day 13","2","4","6","B1") ~ "Baseline",
                             study_event_id %in% c("Day 15","Day 20","Day 22","Day 27","Day 29","Day 34","Day 41","12","14","17","22","A1","A2","A3") ~ "Stress",
                             TRUE ~ "Recovery"))

nback_data$phase <- factor(nback_data$phase, levels = c("Baseline","Stress","Recovery"))

nback_data <- na.omit(nback_data)

#head(nback_data)

nback_data <- nback_data %>% group_by(study, subjectid, phase, variable) %>% filter(subjectid != "6") %>%
    summarize(avg = mean(value, na.rm=TRUE),
              .groups = 'drop') %>%
    group_by(study) %>% mutate(nphases = n_distinct(phase)) %>%
    group_by(study, subjectid) %>% 
    filter(n_distinct(phase) == nphases) %>%
    complete(phase, variable) %>% 
    group_by(study,subjectid,variable) %>%
    mutate(avg,
           Robustness = if_else(variable=="mrt",avg[phase=="Baseline"]-avg[phase=="Stress"],avg[phase=="Stress"]-avg[phase=="Baseline"]),
           Resilience = if_else(variable=="mrt",avg[phase=="Stress"]-avg[phase=="Recovery"],avg[phase=="Recovery"]-avg[phase=="Stress"])) %>%
    group_by(study,subjectid) %>%
    mutate(Stability = ifelse(study=="HIGH",
                              abs(replace_na(avg[variable=="mrt" & phase=="Baseline"]-avg[variable=="mrt" & phase=="Stress"],0)),
                              abs(replace_na(avg[variable=="mrt" & phase=="Baseline"]-avg[variable=="mrt" & phase=="Stress"],0))+abs(replace_na(avg[variable=="mrt" & phase=="Recovery"]-avg[variable=="mrt" & phase=="Stress"],0)))) %>%
    group_by(study,variable) %>%
    mutate(pStability = 1-percent_rank(Stability),
           pRobustness = percent_rank(Robustness),
           pResilience = percent_rank(Resilience))

#subset(nback_data,study=="OPSII")
head(nback_data)

na.omit(dcast(nback_data, study+subjectid+phase~variable, value.var="avg")) %>% group_by(study) %>% summarize(corr = cor(mrt, dprime))

nback_avg_data <- nback_data %>% group_by(study,phase,variable) %>% summarize(mean=mean(avg, na.rm=TRUE),
                                                   .groups = 'drop')
#head(avg_rt_data)
#comined 1-back and 2-back
rr_plot <- ggplot(data = nback_data, aes(x=phase, y=avg, group=subjectid,color=pStability)) + labs(x="Phase",y="Mean Reaction Time (ms)") + labs(color="Stability\nRank\nPercentile") +
        geom_line(linewidth=0.65, alpha=0.7) + geom_point(size=1.0, alpha=0.8)+
        geom_line(data = nback_avg_data, linewidth=1.0, aes(x=phase, y=mean, group=1), colour="#F100F1FF") +
        theme_bw() + ggtitle("Heterogeneity in Nback") +
        scale_color_viridis(labels = scales::percent) +
        facet_grid2(variable ~ study, scales = "free", switch = "y", labeller = labeller(variable = c(`mrt` = "Mean RT", `dprime` = "d prime"))) + 
        #scale_x_discrete(guide = guide_axis(angle = 60)) +
        #scale_color_manual(values = c("#D55E00", "#0072B2", "#B4AB32")) + 
        theme(text = element_text(size=10, face="bold"),
              plot.title = element_text(size=11, hjust = 0.5),
              axis.text.x = element_text(size=7,angle = 55, hjust = 1),
              axis.title.x = element_text(size=11, margin=margin(5,0,0,0)),
              axis.text.y = element_text(size=7, margin=margin(0,2,0,0)),
              axis.title.y = element_blank(), #element_text(size=11, margin=margin(0,0,0,0),vjust = -16),
              legend.text = element_text(size=9, face="plain"),
              legend.margin = margin(0.25,0,0,0, unit="cm"),
              legend.spacing.y = unit(0.1, 'cm'),
              legend.key.height = unit(1,"line"),
              strip.text.x = element_text(margin = margin(.1, 0, .1, 0, "cm")),
              strip.text.y = element_text(margin = margin(0, .1, 0, .1, "cm")),
              strip.placement.y = "outside"
             )
rr_plot
```

```r
## Total Resilience as a function of robustness and resilience (grouped 1-back and 2-back) (same colors for each participant across DV's) based on mrt

options(repr.plot.width=5, repr.plot.height=5, repr.plot.res=600)

nback_data <- nback_agg %>% select(study, subjectid, study_event_id, SessBlockID, mrt, dprime) %>% 
    melt(id.vars = c("study", "subjectid", "study_event_id", "SessBlockID")) %>%
    filter(!(study_event_id %in% c("Day 5","Day 9"))) %>% 
    mutate(phase = case_when(study_event_id %in% c("Day 0","Day 2","Day 4","Day 6","Day 8","Day 10","Day 13","2","4","6","B1") ~ "Baseline",
                             study_event_id %in% c("Day 15","Day 20","Day 22","Day 27","Day 29","Day 34","Day 41","12","14","17","22","A1","A2","A3") ~ "Stress",
                             TRUE ~ "Recovery"))

nback_data$phase <- factor(nback_data$phase, levels = c("Baseline","Stress","Recovery"))

nback_data <- na.omit(nback_data)

#head(nback_data)

nback_data <- nback_data %>% group_by(study, subjectid, phase, variable) %>% filter(subjectid != "6") %>%
    summarize(avg = mean(value, na.rm=TRUE),
              .groups = 'drop') %>%
    group_by(study) %>% mutate(nphases = n_distinct(phase)) %>%
    group_by(study, subjectid) %>% 
    filter(n_distinct(phase) == nphases) %>%
    complete(phase, variable) %>% 
    group_by(study,subjectid,variable) %>%
    mutate(avg,
           Robustness = if_else(variable=="mrt",avg[phase=="Baseline"]-avg[phase=="Stress"],avg[phase=="Stress"]-avg[phase=="Baseline"]),
           Resilience = if_else(variable=="mrt",avg[phase=="Stress"]-avg[phase=="Recovery"],avg[phase=="Recovery"]-avg[phase=="Stress"])) %>%
    group_by(study,subjectid) %>%
    mutate(Stability = ifelse(study=="HIGH",
                              abs(replace_na(avg[variable=="mrt" & phase=="Baseline"]-avg[variable=="mrt" & phase=="Stress"],0)),
                              abs(replace_na(avg[variable=="mrt" & phase=="Baseline"]-avg[variable=="mrt" & phase=="Stress"],0))+abs(replace_na(avg[variable=="mrt" & phase=="Recovery"]-avg[variable=="mrt" & phase=="Stress"],0)))) %>%
    group_by(study,variable) %>%
    mutate(pStability = 1-percent_rank(Stability),
           pRobustness = percent_rank(Robustness),
           pResilience = percent_rank(Resilience))

#subset(nback_data,study=="OPSII")

nback_avg_data <- nback_data %>% group_by(study,phase,variable) %>% summarize(mean=mean(avg, na.rm=TRUE),
                                                   .groups = 'drop')
#head(avg_rt_data)
#comined 1-back and 2-back
rr_plot <- ggplot(data = nback_data, aes(x=phase, y=avg, group=subjectid,color=pStability)) + labs(x="Phase",y="Mean Reaction Time (ms)") + labs(color="Stability\nRank\nPercentile") +
        geom_line(linewidth=0.5, alpha=0.7) + geom_point(size=1.0, alpha=0.8)+
        geom_line(data = nback_avg_data, linewidth=1.0, aes(x=phase, y=mean, group=1), colour="#F100F1FF") +
        theme_bw() + ggtitle("Heterogeneity in Nback") +
        scale_color_viridis(labels = scales::percent) +
        facet_grid2(variable ~ study, scales = "free", switch = "y", labeller = labeller(variable = c(`mrt` = "Mean RT", `dprime` = "d prime"))) + 
        #scale_x_discrete(guide = guide_axis(angle = 60)) +
        #scale_color_manual(values = c("#D55E00", "#0072B2", "#B4AB32")) + 
        theme(text = element_text(size=10, face="bold"),
              plot.title = element_text(size=11, hjust = 0.5,,margin=margin(0,5,5,5)),
              axis.text.x = element_text(size=7,angle = 55, hjust = 1),
              axis.title.x = element_text(size=11, margin=margin(5,0,0,0)),
              axis.text.y = element_text(size=7, margin=margin(0,2,0,0)),
              axis.title.y = element_blank(), #element_text(size=11, margin=margin(0,0,0,0),vjust = -16),
              legend.text = element_text(size=9, face="plain"),
              legend.margin = margin(0.25,0,0,0, unit="cm"),
              legend.spacing.y = unit(0.1, 'cm'),
              legend.key.height = unit(1,"line"),
              strip.text.x = element_text(margin = margin(.1, 0, .1, 0, "cm")),
              strip.text.y = element_text(margin = margin(0, .1, 0, .1, "cm")),
              strip.placement.y = "outside",
              #strip.switch.pad.grid = unit(1, "cm"),
              plot.margin = unit(x=c(0.2,0.2,0.2,0.2),units="cm")
             )
rr_plot
```
